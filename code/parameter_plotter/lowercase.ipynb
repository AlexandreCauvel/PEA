```python
import os

def uppercase_letter_in_filenames(directory_path, target_letter):
    """
    Replaces a specific lowercase letter with its uppercase version 
    in all filenames within a given directory.
    """
    # Safety check: ensure the input is a single lowercase letter
    if len(target_letter) != 1 or not target_letter.islower():
        print("Error: Please provide exactly one lowercase letter (e.g., 'a', 'e', 'x').")
        return

    # Check if the directory exists
    if not os.path.isdir(directory_path):
        print(f"Error: The directory '{directory_path}' does not exist.")
        return

    renamed_count = 0

    # Iterate through all items in the directory
    for filename in os.listdir(directory_path):
        old_filepath = os.path.join(directory_path, filename)
        
        # Process only files (skips subfolders)
        if os.path.isfile(old_filepath):
            if target_letter in filename:
                # Replace the lowercase letter with its uppercase version
                new_filename = filename.replace(target_letter, target_letter.upper())
                new_filepath = os.path.join(directory_path, new_filename)
                
                # Rename the file
                os.rename(old_filepath, new_filepath)
                print(f"Renamed: {filename} ➔ {new_filename}")
                renamed_count += 1

    print(f"\nTask complete! Total files renamed: {renamed_count}")

# --- HOW TO USE IT ---
# 1. Put the path to your folder here (use '.' for the current folder where the script runs)
target_folder = r"C:\Users\Alexa\Desktop\PEA\EIS_data\LDH_EIS_2" 

# 2. Specify the lowercase letter you want to change
letter_to_change = "h" 

# 3. Run the function
uppercase_letter_in_filenames(target_folder, letter_to_change)
```
